```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def rk4(r, t, h):                    #edited; no need for input f
        """ Runge-Kutta 4 method """
        k1 = h*f(r, t)
        k2 = h*f(r+0.5*k1, t+0.5*h)
        k3 = h*f(r+0.5*k2, t+0.5*h)
        k4 = h*f(r+k3, t+h)
        return (k1 + 2*k2 + 2*k3 + k4)/6

def f(r, t):
        a = 2/3
        b = 4/3
        c = 1
        d = 1
        x, y = r[0], r[1]
        fxd = a*x - b*x*y
        fyd = c*x*y - d*y
        return np.array([fxd, fyd], float)

h=0.001                               #edited
tpoints = np.arange(0, 12, h)         #edited
xpoints, ypoints  = [], []
r = np.array([0.9, 0.9], float)
for t in tpoints:
        xpoints.append(r[0])          #edited
        ypoints.append(r[1])          #edited
        r += rk4(r, t, h)             #edited; no need for input f
plt.plot(xpoints, ypoints)
plt.xlabel("x")
plt.ylabel("y")
plt.show()

tarr = np.asarray(tpoints)
xarr = np.asarray(xpoints)
yarr = np.asarray(ypoints)

points = np.vstack((tarr, xarr, yarr))

# with open('LV 2,3 4,3 1 1 0.9 0.9 0 12 0.0001.npy', 'wb') as f:

    # np.save(f, points)
```

```python
num_models = 4

n_total = len(points[0])
n_model = int(n_total/num_models)
```
